```python

# ============================================================
# TASK 1: DATA IMPORT, EXPLORATION AND STATISTICAL ANALYSIS
# Project: Electric Vehicle Data Analysis Using Python
# ============================================================

# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings
import zipfile

from scipy.stats import shapiro
from google.colab import files
from IPython.display import display

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

os.makedirs("EV_Results", exist_ok=True)

print("=" * 60)
print("ELECTRIC VEHICLE DATA ANALYSIS")
print("=" * 60)


# 2. IMPORT DATASET
url = "https://data.wa.gov/resource/f6w7-q2d2.csv?$limit=50000"

try:
    df = pd.read_csv(url)
    print("\nDataset downloaded successfully!")

except Exception:
    print("Upload your EV CSV dataset:")
    uploaded = files.upload()
    filename = next(iter(uploaded))
    df = pd.read_csv(filename)

print("\nFIRST FIVE RECORDS")
display(df.head())


# 3. DATASET DESCRIPTION
print("\nDATASET DESCRIPTION")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Column Names:", df.columns.tolist())

print("\nDATASET INFORMATION")
df.info()

print("\nSTATISTICAL DESCRIPTION")
display(df.describe(include="all").T)


# 4. DATA INSPECTION
print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nDUPLICATE RECORDS")
print(df.duplicated().sum())

print("\nDATA TYPES")
print(df.dtypes)


# 5. DATA PREPROCESSING
df.drop_duplicates(inplace=True)

num_cols = df.select_dtypes(include=np.number).columns.tolist()
cat_cols = df.select_dtypes(include=["object", "category"]).columns.tolist()

for col in num_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

for col in cat_cols:
    if df[col].isnull().any():
        mode = df[col].mode()
        df[col] = df[col].fillna(
            mode.iloc[0] if len(mode) else "Unknown"
        )

df.dropna(axis=1, how="all", inplace=True)

print("\nPREPROCESSING COMPLETED")
print("Remaining missing values:", df.isnull().sum().sum())
print("Remaining duplicates:", df.duplicated().sum())
print("Cleaned shape:", df.shape)


# 6. SELECT IMPORTANT FEATURES
def find_column(keyword):
    for col in df.columns:
        if keyword.lower() in str(col).lower():
            return col
    return None

range_col = find_column("electric_range")
year_col = find_column("model_year")
price_col = find_column("base_msrp")
make_col = find_column("make")
type_col = find_column("electric_vehicle_type")

excluded = [
    "vin", "vehicle id", "dol_vehicle_id",
    "postal", "zip", "census", "latitude",
    "longitude", "district", "tract"
]

analysis_cols = [
    c for c in num_cols
    if c in df.columns
    and not any(x in c.lower() for x in excluded)
]

print("\nSELECTED NUMERICAL FEATURES:", analysis_cols)
print("SELECTED CATEGORICAL FEATURES:", cat_cols)


# 7. STATISTICAL ANALYSIS
print("\nDESCRIPTIVE STATISTICAL ANALYSIS")

if not analysis_cols:
    raise ValueError("No numerical attributes available.")

stats = pd.DataFrame({
    "Mean": df[analysis_cols].mean(),
    "Median": df[analysis_cols].median(),
    "Variance": df[analysis_cols].var(),
    "Standard Deviation": df[analysis_cols].std(),
    "Minimum": df[analysis_cols].min(),
    "Maximum": df[analysis_cols].max(),
    "Range": df[analysis_cols].max() - df[analysis_cols].min()
})

display(stats.round(3))

stats.to_csv("EV_Results/Statistical_Summary.csv")


# 8. COVARIANCE
print("\nCOVARIANCE MATRIX")

covariance = df[analysis_cols].cov()
display(covariance.round(3))

covariance.to_csv("EV_Results/Covariance.csv")


# 9. CORRELATION
print("\nCORRELATION MATRIX")

correlation = df[analysis_cols].corr()
display(correlation.round(3))

correlation.to_csv("EV_Results/Correlation.csv")


# 10. PROBABILITY DISTRIBUTION
print("\nPROBABILITY DISTRIBUTION ANALYSIS")

selected = (
    range_col if range_col in analysis_cols
    else analysis_cols[0]
)

data = df[selected].dropna()

print("Selected Attribute:", selected)
print("Mean:", data.mean())
print("Median:", data.median())
print("Variance:", data.var())
print("Standard Deviation:", data.std())
print("Skewness:", data.skew())
print("Kurtosis:", data.kurtosis())

median = data.median()

print("P(Value > Median):", (data > median).mean())
print("P(Value < Median):", (data < median).mean())

sample = data.sample(min(5000, len(data)), random_state=42)

if len(sample) >= 3:
    statistic, p_value = shapiro(sample)
    print("Shapiro-Wilk Statistic:", statistic)
    print("P-value:", p_value)

    if p_value < 0.05:
        print("Distribution: Evidence of non-normality")
    else:
        print("Distribution: Normality not rejected")


# 11. HISTOGRAM
plt.figure(figsize=(9, 5))
sns.histplot(data=data, bins=30, kde=True)
plt.title(f"Histogram of {selected}")
plt.xlabel(selected)
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("EV_Results/Histogram.png")
plt.show()


# 12. BOX PLOT
plt.figure(figsize=(9, 5))
sns.boxplot(x=data)
plt.title(f"Box Plot of {selected}")
plt.xlabel(selected)
plt.tight_layout()
plt.savefig("EV_Results/Boxplot.png")
plt.show()


# 13. SCATTER PLOT
if range_col in analysis_cols and price_col in analysis_cols:
    x_col, y_col = range_col, price_col
elif len(analysis_cols) >= 2:
    x_col, y_col = analysis_cols[:2]
else:
    x_col = y_col = None

if x_col:
    plt.figure(figsize=(9, 5))
    sns.scatterplot(data=df, x=x_col, y=y_col, alpha=0.5)
    plt.title(f"{x_col} vs {y_col}")
    plt.tight_layout()
    plt.savefig("EV_Results/Scatterplot.png")
    plt.show()


# 14. CORRELATION HEATMAP
plt.figure(figsize=(12, 7))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig("EV_Results/Heatmap.png")
plt.show()


# 15. MANUFACTURER ANALYSIS
if make_col:
    top_makes = df[make_col].value_counts().head(10)

    print("\nTOP 10 MANUFACTURERS")
    display(top_makes)

    plt.figure(figsize=(11, 5))
    sns.barplot(x=top_makes.index, y=top_makes.values)
    plt.title("Top 10 Electric Vehicle Manufacturers")
    plt.xlabel("Manufacturer")
    plt.ylabel("Vehicle Count")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig("EV_Results/Manufacturer.png")
    plt.show()


# 16. VEHICLE TYPE ANALYSIS
if type_col:
    vehicle_types = df[type_col].value_counts()

    print("\nVEHICLE TYPE DISTRIBUTION")
    display(vehicle_types)

    plt.figure(figsize=(9, 5))
    vehicle_types.plot(kind="bar")
    plt.title("Electric Vehicle Type Distribution")
    plt.xlabel("Vehicle Type")
    plt.ylabel("Vehicle Count")
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.savefig("EV_Results/Vehicle_Type.png")
    plt.show()


# 17. MODEL YEAR ANALYSIS
if year_col:
    years = df[year_col].value_counts().sort_index()

    plt.figure(figsize=(11, 5))
    sns.lineplot(x=years.index, y=years.values, marker="o")
    plt.title("Electric Vehicles by Model Year")
    plt.xlabel("Model Year")
    plt.ylabel("Vehicle Count")
    plt.tight_layout()
    plt.savefig("EV_Results/Model_Year.png")
    plt.show()


# 18. OUTLIER ANALYSIS
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data[(data < lower) | (data > upper)]

print("\nOUTLIER ANALYSIS")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
print("Number of Outliers:", len(outliers))


# 19. AUTOMATIC INSIGHTS
print("\n" + "=" * 60)
print("DATA ANALYSIS FINDINGS")
print("=" * 60)

print("Total Vehicles:", len(df))
print("Total Features:", len(df.columns))

if make_col:
    print("Most Common Manufacturer:",
          df[make_col].mode().iloc[0])

if type_col:
    print("Most Common EV Type:",
          df[type_col].mode().iloc[0])

if range_col in analysis_cols:
    print("Average Electric Range:", df[range_col].mean())
    print("Maximum Electric Range:", df[range_col].max())
    print("Minimum Electric Range:", df[range_col].min())

if len(analysis_cols) >= 2:
    pairs = correlation.where(
        np.triu(np.ones(correlation.shape), k=1).astype(bool)
    ).stack()

    if len(pairs):
        pair = pairs.abs().idxmax()
        print("Strongest Correlation:", pair)
        print("Correlation Value:", pairs.loc[pair])

print("\nINTERPRETATION:")
print("1. Statistical measures describe the EV dataset.")
print("2. Histogram explains numerical distributions.")
print("3. Boxplot identifies possible outliers.")
print("4. Scatterplot displays relationships.")
print("5. Heatmap identifies correlations.")
print("6. Bar charts compare categorical attributes.")


# 20. SAVE FINAL RESULTS
df.to_csv("EV_Results/Cleaned_EV_Dataset.csv", index=False)

report = f"""
ELECTRIC VEHICLE DATA ANALYSIS REPORT

Project: Exploratory Data Analysis of EV Data Using Python

Total Records: {len(df)}
Total Attributes: {len(df.columns)}

Numerical Features:
{analysis_cols}

Categorical Features:
{cat_cols}

STATISTICAL SUMMARY:
{stats.to_string()}

COVARIANCE MATRIX:
{covariance.to_string()}

CORRELATION MATRIX:
{correlation.to_string()}

PROBABILITY DISTRIBUTION:
Selected Feature: {selected}
Mean: {data.mean()}
Median: {data.median()}
Standard Deviation: {data.std()}
Skewness: {data.skew()}

OUTLIER ANALYSIS:
Number of Outliers: {len(outliers)}

CONCLUSION:
The dataset was imported, inspected, cleaned and analyzed.
Statistical measures and visualizations were generated to
understand EV data distributions and relationships.
"""

with open("EV_Results/Analysis_Report.txt", "w") as f:
    f.write(report)

# ZIP all generated files
import shutil

shutil.make_archive("EV_Analysis_Submission", "zip", "EV_Results")

print("\n" + "=" * 60)
print("TASK 1 COMPLETED SUCCESSFULLY!")
print("=" * 60)

print("\nGenerated Files:")
for file in os.listdir("EV_Results"):
    print(file)

files.download("EV_Analysis_Submission.zip")
```
